## Evaluation design

Microcalcifications are nested within patients. We therefore split patient IDs first and keep every measurement from one patient in a single partition. A microcalcification model is fitted on training patients, the aggregation threshold is selected on validation patients, and both tasks are evaluated once on test patients.


```python
from breast_mc import run_study, synthetic_microcalcifications

frame = synthetic_microcalcifications()
frame.shape
```

This notebook uses synthetic hierarchical feature vectors only to verify the software. Its metrics are not clinical results and do not reproduce the private course dataset.
